Pydantic models: Creating and using models

=> Pydantic models are foundation of data validation in Python. They use python type annotations to define the structure and validate data at runtime.

```python
#pydantic basics: creating and using models

from pydantic import BaseModel
from dataclasses import dataclass
```

```python
#here, since we're using pydantic model of BaseModel, we're able to do validation of given, that's the reason when we gave wrong type, it gave validation error here!
class Person(BaseModel):
    name:str
    age:int
    city:str

person=Person(name="Kris",age=21,city="Hyd")
person1=Person(name="Kris",age=21,city=12)
print(person)
print(type(person))
print(person1)
print(type(person1))
```

```python
"""
instead of using BaseModel, used dataClass here, still working fine, butttttt...
here, when we gave for person1, the city as 12(integer), instead of string, it is not erroring it out, as BaseModel(ValidationError: 1 validation error for Person)
so, here it is not validating, just giving as is!!!
"""
@dataclass
class Person():
    name:str
    age:int
    city:str

person=Person(name="Kris",age=21,city="Hyd")
person1=Person(name="Kris",age=21,city=12)
print(person)
print(type(person))
print(person1)
print(type(person1))
```

2. Model with optional fields!

=> Add optional fields using Python's Optional Type!

```python
"""
2. Let's see another one: 
"""

from typing import Optional

class Employee(BaseModel):
    id:str
    name:str
    dept:str
    salary:Optional[float]=None #here, giving optional with default value as None
    is_active:Optional[bool]=False #here, giving optional with default value as True
```

```python
emp1=Employee(id='1',name='Kris',dept='CSE')
print(emp1)
emp1=Employee(id='2',name='Lucy',dept='CSE',salary=100000.00,is_active=True)
print(emp1)
```

```python
from pydantic import BaseModel
from typing import List

class Class(BaseModel):
    room_no:str
    students:List[str]
    capacity:int
```

```python
cls=Class(room_no="12",students=["sai","sam"],capacity=2)
print(cls)
#even typecasting is possible, if you give tuple in students, since we gave list, it will typecast!
cls2=Class(room_no="12",students=("sai","sam"),capacity=2)
print(cls2)
```

```python
try:
    invalid_cls=Class(room_no="12",students=32,capacity=2)
except ValueError as e:
    print(e)
```

3. Model with nested models

=> create complex structures with nested models

```python
from pydantic import BaseModel

class Address(BaseModel):
    street:str
    city:str
    zip_code:str

class Customer(BaseModel):
    cust_id:int
    name:str
    address:Address

#create a customer with nested address!
cust1=Customer(
    cust_id=1,
    name="kris",
    address={"street":"puli","city":"kng","zip_code":"523230"}
)
print(cust1)
```

4. Pydantic fields: Customizations and constraints

=> The Field function in Pydantic enhance model fields beyond basic type hints by allowing you to specify validation rules, default values, aliases and more. 
Here's a comprehensive tutorial with examples!!!

```python
from pydantic import BaseModel,Field

class Item(BaseModel):
    name:str=Field(min_length=2,max_length=50)
    price:float=Field(gt=0,le=1000) #greater than 0 and less than or equal to 1000
    qty:int=Field(ge=0) #greater than equal to zero
    
item=Item(name="kal",price=200.0,qty=1)
print(item)
#if we give beyond the field constriants, let's see!!!
item2=Item(name="Lucy",price=2005.0,qty=-1)
print(item2)
```

```python
from pydantic import BaseModel,Field

class User(BaseModel):
    name:str=Field(...,description="Unique username")
    age:int=Field(default=18,description="Defaulted age to 18")
    email:str=Field(default_factory=lambda:"user@example.com",description="Default email address")

user1=User(name="Alice")
print(user1)
user2=User(name="Bob",age=21,email="bob@gmail.com")
print(user2)
```

```python
print(User.schema())
```
