```python
class BankAcc:
    def __init__(self, owner, acc_no, balance):
        self.owner=owner
        self.acc_no=acc_no
        self.balance=balance

    def deposit(self, amount):
        self.balance+=amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount>self.balance:
            print("Insufficient funds.")
        else:
            self.balance-=amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
    def balance_enquiry(self):
        print(f"Current balance: {self.balance}")
            
acc1=BankAcc("Alice", "123456", 1000)
acc1.deposit(500)
acc1.withdraw(200)
acc1.balance_enquiry()
```

```python
#parent class
class Car:
    def __init__(self,gears,engine):
        self.gears=gears
        self.engine=engine
    def drive(self):
        print(f"The person will drive '{self.engine}' car")
```

```python
tata=Car(5,"Revotron(petrol)")
tata.drive()
```

```python
#child class
class Tesla(Car):
    def __init__(self,gears,engine,isSelfDrive):
        super().__init__(gears,engine)
        self.isSelfDrive=isSelfDrive
    def checkSelfDrive(self):
        return self.isSelfDrive
    
ramu=Tesla(6,"Cyberbeast",True)
print(ramu.checkSelfDrive())
rams=Tesla(6,"Beast",False)
print(rams.checkSelfDrive())
```

```python
#multiple inheritance(when cls inherits from more than one base cls)
#base cls1
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print("subclass must implement this method!!!")
        
#base cls2
class Pet:
    def __init__(self,owner):
        self.owner=owner

#inherit both cls(derived cls)
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    def print(self):
        return f" owner and name: {self.owner} {self.name}"
    
d=Dog("snoopy","bob")
d.print()
```

```python
#method overriding

class Animal:
    def speak(self):
        return "sound of animal"
    
class Dog(Animal):
    def speak(self):
        return "Dog sound"

class Cat(Animal):
    def speak(self):
        return "cat meow"
    
dog=Dog()
cat=Cat()
animal=Animal()
print(dog.speak())
print(cat.speak())
print(animal.speak())
```

```python
#using abstract cls
from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
    
class Car(Vehicle):
    def start(self):
        return "Car started"

class Bike(Vehicle):
    def start(self):
        return "Bike started"

def start_vehicle(vehicle):
    print(vehicle.start())    

car=Car()
bike=Bike()

start_vehicle(car)
start_vehicle(bike)
```

```python
#protected variables
class Person:
    def __init__(self,name,age,gender):
        self._name=name
        self._age=age
        self._gender=gender

class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)
        
emp=Employee("Kris",21,"M")
print(emp._name)
```

```python
#encapsulation with getter and setter methods
class Person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
    def get_name(self):
        return self.__name
    def set_name(self,name):
        self.__name=name
        
class Employee(Person):
    def __init__(self,name,age):
        super().__init__(name,age)
        
emp=Employee("Lucy",21)
emp.get_name()
```

```python
#abstraction
from abc import ABC,abstractmethod

#abstract base cls
class Vehicle(ABC):
    def drive(self):
        print("Vehicle for drive")
    
    @abstractmethod
    def start_engine(self):
        pass
    
class Car(Vehicle):
    def start_engine(self):
        print("Car engine start")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()
    
car=Car()
operate_vehicle(car)
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f"{self.name}, {self.age}years old"
    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"

ps1=Person("Kris",21)
print(ps1)
print(repr(ps1))
```
