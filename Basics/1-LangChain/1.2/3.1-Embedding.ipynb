```python
#instead of this vector, i can define my own dimension as follows

embeddings_1024=GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001",dimensions=1024)
embeddings_1024
```

```python
text="This is a tutorial on GoogleAI embedding"
query_result=embeddings_1024.embed_query(text)
query_result
```

```python
print(len(query_result))
```

```python
#Text Loader
from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
textDocs=loader.load()
textDocs
```

```python
#convert pdf to text

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_docs=text_splitter.split_documents(textDocs)
final_docs
```

```python
#now, i splitted the docs, now i want to convert these docs to vector and store in vector db
# Here we can use Vector Embedding and Vector StoreDB

from langchain_community.vectorstores import Chroma

#1. to convert all the docs to vector
db=Chroma.from_documents(final_docs,embeddings_1024)
db
```

```python
#2. Retrieve results from query vectorstore db
query="While normally used to facilitate communication with others"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```
