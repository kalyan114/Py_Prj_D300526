#### Chroma
Chroma is a AI-native-open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0

```python
#building sample vectordb
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
data=loader.load()
data
```

```python
#split
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=0)
splits=text_splitter.split_documents(data)
splits
```

```python
embedding=OllamaEmbeddings(model='nomic-embed-text')
vectordb=Chroma.from_documents(splits,embedding=embedding)
vectordb
```

```python
#query it 
query="While normally used to facilitate communication with others, do people may also use speech without the intent to communicate?"
docs=vectordb.similarity_search(query)
```

```python
docs[0].page_content
```

```python
#saving to the disk
vectordb=Chroma.from_documents(documents=splits,embedding=embedding,persist_directory="./chroma_db")
```

```python
#load from disk
db2=Chroma(persist_directory='./chroma.db',embedding_function=embedding)
docs=db2.similarity_search(query)
docs
# print(docs[0].page_content)
```

```python
#similarity search with score
docs=vectordb.similarity_search_with_score(query)
docs
```

```python
#Retriever option
retriever=vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
