- Get setup with LangChain, LangSmith and LangServe

- Trace your application with LangSmith

```python
#let's import all the details
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GEMINI_API_KEY']=os.getenv('GEMINI_API_KEY')
#langsmith tracking
os.environ['LANGCHAIN_API_KEY']=os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_TRACING_V2']="true"
os.environ['LANGCHAIN_PROJECT']=os.getenv('LANGCHAIN_PROJECT')
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model='gemini-2.5-flash')
print(llm)
```

```python
#chatprompt templates

from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer. Provide me answers based on the question"),
        ("user","{input}")
    ]
)

prompt
```

```python
chain=prompt|llm

res=chain.invoke({"input":"Can you tell me about Langsmith?"})
print(res)
```

```python
print(res.content)
```

```python
#to use output parser to display our response msg, so we can use "{stroutput parser}"
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()

chain=prompt|llm|output_parser

response=chain.invoke({"input":"Can you tell me about Langchain?"})
print(response)
```
